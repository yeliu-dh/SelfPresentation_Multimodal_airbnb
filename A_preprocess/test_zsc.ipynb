```python
DATA_FOLDER="../data_processed"
dict_input={}

for folder in os.listdir(DATA_FOLDER):
    files=os.listdir(os.path.join(DATA_FOLDER, folder))
    for f in files:
        if f.startswith('listings'):
            path_listings=os.path.join(DATA_FOLDER, folder, f)
            if "processed" in f:
                path_listings_zsc=os.path.join(DATA_FOLDER, folder, f.replace("processed","zsc"))
            else : 
                path_listings_zsc=os.path.join(DATA_FOLDER, folder, f"listings_zsc_{folder}.csv")

    path_results=os.path.join(DATA_FOLDER, folder, f"resultsZSC_{folder}.csv")
    
    
    dict_input[folder]={"listings":path_listings,
                       "results":path_results,
                       "listings_zsc":path_listings_zsc}

for k,v in dict_input.items():
    print(k)
    path_listings=v['listings']
    path_results=v['results']
    path_listings_zsc=v['listings_zsc']

    print(path_listings)
    print(path_results)
    print(path_listings_zsc,"\n")
  
```
